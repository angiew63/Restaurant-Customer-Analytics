# Restaurant Customer Analytics — Exploratory Data Analysis

## Business Context

ChefMoz aims to improve customer satisfaction and restaurant performance by understanding customer preferences, dining behaviors, and rating patterns.

This exploratory data analysis investigates trends, relationships, and distributions within the cleaned restaurant dataset to uncover insights that can guide strategic improvements.

The analysis focuses on identifying:

- Customer demographic patterns
- Dining preference trends
- Restaurant service performance
- Rating behavior across customer groups
- Factors associated with customer satisfaction

## Objective

The objective of this notebook is to extract actionable insights from the cleaned dataset through visual and statistical exploration.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

### Load Clean Dataset

```python
df = pd.read_csv("../data/processed/chefmoz_clean.csv")

df.head()
```

### 1. Ratings Distribution

```python
sns.countplot(data=df, x='rating')
plt.title("Distribution of Overall Ratings")
plt.show()
```

### Insight

The overall rating distribution shows the general pattern of customer satisfaction across restaurants.

The concentration of responses around the dominant rating category suggests that customer experiences are clustered rather than evenly distributed, indicating consistent satisfaction trends across most restaurant visits.

### 2. Food Rating

```python
sns.countplot(data=df, x='food_rating')
plt.title("Food Rating Distribution")
plt.show()
```

### Insight

Food rating distribution highlights customer perceptions of food quality.

A higher concentration of positive food ratings would suggest that food quality is a major strength, while wider spread across categories may indicate inconsistency in food experiences across restaurants.

### 3. Service Rating

```python
sns.countplot(data=df, x='service_rating')
plt.title("Service Rating Distribution")
plt.show()
```

### Insight

Service ratings reveal customer perceptions of restaurant service quality.

Differences between service and food rating distributions may indicate whether customer satisfaction is driven more strongly by food quality or service experience.

### 4. Customer Budget Distribution

```python
sns.countplot(data=df, x='budget')
plt.title("Customer Budget Categories")
plt.show()
```

### Insight

The budget distribution indicates the dominant spending profile of customers in the dataset.

A concentration within a specific budget category suggests that restaurant offerings may primarily cater to a particular economic segment.

### 5. Marital Status

```python
sns.countplot(data=df, x='marital_status')
plt.title("Customer Marital Status")
plt.show()
```

### Insight

The customer base is primarily concentrated within specific marital status groups.

This distribution provides context for understanding whether household structure may influence dining preferences and restaurant rating behavior.

### 6. Demographic Analysis

#### Height Distribution

```python
sns.histplot(df['height'], kde=True)
plt.title("Customer Height Distribution")
plt.show()
```

### Insight

The height distribution appears to follow a natural demographic spread, suggesting no major anomalies after preprocessing.

This confirms consistency in physical attribute records.

#### Weight Distribution

```python
sns.histplot(df['weight'], kde=True)
plt.title("Customer Weight Distribution")
plt.show()
```

### Insight

The weight distribution shows the spread of customer physical characteristics and confirms that preprocessing successfully addressed potential outliers or unrealistic values.

###  7. Budget vs Rating

```python
sns.boxplot(data=df, x='budget', y='rating')
plt.title("Budget vs Customer Rating")
plt.show()
```

### Insight

The relationship between budget category and customer ratings helps assess whether spending power influences restaurant satisfaction.

Differences in median ratings across budget groups may indicate varying expectations and service perceptions among customer segments.

### 8. Marital Status vs Rating

```python
sns.boxplot(data=df, x='marital_status', y='rating')
plt.title("Marital Status vs Rating")
plt.show()
```

### Insight

Rating distributions across marital status categories reveal whether household structure is associated with restaurant satisfaction patterns.

Minimal variation would suggest that marital status has limited influence on customer rating behavior.

### 9. Ambience vs Rating

```python
sns.boxplot(data=df, x='ambience', y='rating')
plt.xticks(rotation=45)
plt.title("Restaurant Ambience vs Rating")
plt.show()
```

### Insight

Customer ratings vary across ambience preferences, indicating that dining environment plays a role in perceived restaurant satisfaction.

Higher median ratings for specific ambience categories suggest stronger alignment between customer environmental preferences and restaurant experience.

### 10. Transport vs Rating

```python
sns.boxplot(data=df, x='transport', y='rating')
plt.xticks(rotation=45)
plt.title("Transport Mode vs Rating")
plt.show()
```

### Insight

Transport mode provides insight into accessibility-related satisfaction patterns.

Differences in rating distributions may suggest that convenience of restaurant access influences overall customer experience.

### Correlation Analysis

```python
numeric_cols = df.select_dtypes(include=np.number)

sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

### Insight

The correlation heatmap highlights linear relationships among numerical variables.

Strong correlations between rating-related variables may indicate shared drivers of satisfaction, while weak correlations suggest that customer satisfaction is influenced more by categorical behavioral and preference-based factors.

# Key Insights

Key observations from the analysis include:

- Most customers gave moderate restaurant ratings
- Food quality appears more influential than service quality
- Budget category shows observable differences in rating behavior
- Ambience preferences vary across customer groups
- Certain demographic attributes show weak relationships with ratings
