# Executive Summary

This project developed a machine learning-driven restaurant recommendation and customer segmentation framework using customer demographics, dining preferences, restaurant characteristics, and behavioral indicators.

The analysis combined predictive modeling and customer segmentation techniques to identify the key drivers of restaurant ratings and uncover distinct customer groups.

Key outcomes include:

- Identification of the strongest factors influencing restaurant satisfaction.
- Development of predictive models capable of estimating customer ratings.
- Discovery of ten customer segments with different dining preferences and satisfaction patterns.
- Generation of actionable insights that can improve recommendation quality, customer engagement, and restaurant performance.

The findings demonstrate that customer satisfaction is strongly influenced by cuisine compatibility, dining experience quality, service performance, and value for money.

# Business Problem

Restaurant customers have diverse preferences, spending habits, and dining expectations.

Traditional recommendation approaches often fail to account for these differences, resulting in:

- Poor customer experiences
- Lower satisfaction ratings
- Reduced customer retention
- Missed revenue opportunities

The objective of this project was to develop data-driven methods to:

1. Predict customer restaurant ratings.
2. Identify the factors driving customer satisfaction.
3. Segment customers into meaningful groups.
4. Generate personalized recommendations and targeted business strategies.

# Key Insights

The exploratory analysis revealed several important patterns:

### 1. Customer Preferences Matter

Customers tend to provide higher ratings when restaurant offerings align with their cuisine preferences.

### 2. Dining Experience Drives Satisfaction

Food quality and service quality are strongly associated with overall restaurant ratings.

### 3. Budget Influences Expectations

Customers with different budget levels evaluate restaurants differently, suggesting varying expectations regarding value for money.

### 4. Social and Family Factors Matter

Dining behavior is influenced by social activities and family-oriented preferences, indicating opportunities for targeted restaurant experiences.

### 5. Customer Profiles Are Diverse

The customer base is heterogeneous, supporting the need for personalized recommendation systems rather than one-size-fits-all approaches.

# Predictive Model Results Summary

Several machine learning models were evaluated to predict customer ratings.

## Baseline Models

| Model | Performance |
|---------|---------|
| Linear Regression | Baseline benchmark |
| Decision Tree | Moderate performance |
| Random Forest | Strong performance |

## Advanced Models

| Model | Performance |
|---------|---------|
| Gradient Boosting | Improved accuracy |
| XGBoost | Strong predictive capability |
| Stacking Ensemble | Best overall performance |

## Key Finding

Ensemble methods consistently outperformed simpler models because they captured complex relationships between customer preferences and restaurant characteristics.

The final selected model achieved the highest predictive performance and provides a reliable foundation for restaurant recommendation systems.

# Customer Segmentation Summary

Clustering analysis identified ten distinct customer groups.

## Major Segments Identified

### Highly Satisfied Premium Diners
- Highest ratings across food and service
- Strong value perception
- Ideal loyalty program candidates

### Cuisine-Focused Customers
- Satisfaction driven primarily by cuisine compatibility
- Highly responsive to personalized recommendations

### Family-Oriented Diners
- Value family-friendly dining experiences
- Price-sensitive but loyal when needs are met

### Social Experience Seekers
- Prioritize social dining opportunities
- Require enhanced customer experiences

### Budget-Conscious Customers
- Sensitive to affordability and value
- Responsive to promotions and discounts

### Dissatisfied Customers
- Lowest satisfaction scores
- Represent key improvement opportunities
